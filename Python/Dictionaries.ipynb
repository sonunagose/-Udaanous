# Dictionaries

```python
Student = {
    "name" : "Ram",
    "Age" : 35,
    "Grade" : "A"
}
print(Student)
```

# Dictionaries Slicing

```python
Company_Profile = {
    "Name" : "Shayam",
    "Age" : 40,
    "City" : "Nagpur",
    "mob" : 123456789
}
Name = Company_Profile["Name"]
print(Name)
Age = Company_Profile["Age"]
print(Age)
City = Company_Profile["City"]
print(City)
mob = Company_Profile["mob"]
print(mob)
```

# Value Add/Remove

## ADD Value

```python
Company_Profile["email"] = "shayam12@gmail.com"
print(Company_Profile)
```

```python
Company_Profile["Dapartment"] = ["HR", "Sales"]
print(Company_Profile)
```

## Remove Value

```python
del Company_Profile["mob"]
print(Company_Profile)
```

# Dictionaries Extractions

## Key

```python
key = Company_Profile.keys()
print(key)
```

## Value

```python
value = Company_Profile.values()
print(value)
```

## Items

```python
items = Company_Profile.items()
print(items)
```
