# Branching

```python
Age = 25
if Age >= 18:
    print("Eligible for Vote")
else:
    print("Not Eligible for Vote")
```

### If Statement

```python
Number = 7 
if Number % 2 == 1:
    print("This is odd Number")
```

### If-else Statement

```python
Marks = 80
if Marks >= 60:
    print("Student is Pass")
else:
    print("Student is Fail")
```

### If-elif-else Statement

```python
Scores = int(input("Enter the number"))
if Scores >= 90:
    print("Grade A")
elif Scores >= 75:
    print("Grade B")
else:
    print("Grade C")
    
```

### If – else Statement + or

```python
Fruits = 100
if Fruits < 10 or Fruits > 90:
    print("Fruits is Good")
else:
    print("Fruits is Not Good")
```

### If – else Statement + and

```python
Age = 18
Income = 10000
if Age > 18 and Income > 10000:
    print("Eligible for Loan")
else:
    print("Not Eligible For Loan")
```
