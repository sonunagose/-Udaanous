# Forward Indexing

```python
my_string = "Anaconda"
print(my_string[1])
print(my_string[3])
print(my_string[4])
```

# Backward Indexing

```python
my_string = "Anaconda"
print(my_string[-1])
print(my_string[-2])
print(my_string[-4])
```

# Slicing Method for Strings

```python
my_string = "Mathmatics"
sliced = my_string[0:6]
print(sliced)
```

# Striding Method for Strings

```python
my_string = "Mathmatics"
strided = my_string[0:10:3]
print(strided)
```

# Len() Function for Strings

```python
my_string = "Mathmatics"
print(len(my_string))
```

# Concatenation of Strings

```python
string_1 = "Science"
string_2 = "Math"
result = string_1 + " " + string_2
print(result)
```

# Escape Sequence in Strings

```python
print("Good/nMorning")
```

```python
print("Python\tAnaconda")
```

# Upper() & Lower() for Strings

```python
my_string = "Programming"
print(my_string.upper())
print(my_string.lower())
```

# Replace() Function for Strings

```python
my_string = " I Love Jupyter"
replaced = my_string.replace("Jupyter" ,"Python")
print(replaced)
```
