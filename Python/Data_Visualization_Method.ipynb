# 23.Bar Chart

```python
# Using Matplot
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'A', 'C', 'B', 'C', 'A']
}
df = pd.DataFrame(data)

freq_churned = df['Product'].value_counts()

plt.figure(figsize=(8, 6))

freq_churned.plot(kind='bar',color=['Green'])

plt.title('Frequency of Product')
plt.xlabel('Product')
plt.ylabel('Frequency')
plt.show()
```

```python
# Using Seaborn

import seaborn as sns
import pandas as pd

data = {
    'Product_category': ['A', 'B', 'C', 'A', 'B', 'A', 'C', 'B', 'C', 'A']
}
df = pd.DataFrame(data)

sns.countplot(x='Product_category', data=df)

plt.title('Frequency of Categories')
plt.xlabel('Product_category')
plt.ylabel('Frequency')
plt.show()
```

# 24.Stack Or Cluster Chart

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample data
data = {
    'Product_1': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'A'],
    'Product_2': ['X', 'Y', 'X', 'Y', 'Z', 'X', 'Y', 'Z']
}
df = pd.DataFrame(data)

cross_tab = pd.crosstab(df['Product_1'], df['Product_2'])


plt.figure(figsize=(10, 6))                  
cross_tab.plot(kind='bar', stacked=False)    
plt.title('The title of the graph')         
plt.xlabel('The label of the x-axis')        
plt.ylabel('The label of the y-axis')        
plt.show()                                 
```

```python
# Using Seaborn

import seaborn as sns
import pandas as pd

data = {
    'Product_1': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'A'],
    'Product_2': ['X', 'Y', 'X', 'Y', 'Z', 'X', 'Y', 'Z']
}
df = pd.DataFrame(data)

sns.countplot(x='Product_1', hue='Product_2', data=df) 
plt.title('The title of the graph')                      
plt.xlabel('The label of the Product_1')                    
plt.ylabel('The label of the Product_2')                   
plt.show()                                               
```

# 25.Pie Chart

```python
import pandas as pd

percentage = pd.Series([80, 50, 100], index=['Company 1', 'Company 2', 'Company 3'])

plt.figure(figsize=(6, 6))
percentage.plot(kind='pie', autopct='%1.2f%%')
plt.title('Sales Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

# 26.Line Plot

```python
# Using Matplot
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Product': ['A', 'B', 'C', 'D', 'E'],
    'Value': [5, 10, 20, 25, 30]
}
df = pd.DataFrame(data)
group_by = df.groupby('Product')['Value'].sum()

plt.figure(figsize=(10, 6)) 
group_by.plot(kind='line', marker='o')  
plt.title('The title of the graph')  
plt.xlabel('The label of the x-axis')  
plt.ylabel('The label of the y-axis')  
plt.show()  
```

```python
# Using Seaborn (with sns.lineplot)

import seaborn as sns
import pandas as pd

# Sample data
data = {
    'Product': ['A', 'B', 'C', 'D', 'E'],
    'Value': [5, 10, 20, 25, 30]
}
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))  
sns.lineplot(x='Product', y='Value', data=df, marker='o') 
plt.title('The title of the graph')  
plt.xlabel('The label of the x-axis')  
plt.ylabel('The label of the y-axis')  
plt.show() 

```

# 27.Histogram

```python
# Using matplotlib

import matplotlib.pyplot as plt
import pandas as pd

# Sample data
data = {
    'Number': [5,10,15,20,25,30,35,40,45,50]
}
df = pd.DataFrame(data)

# Plotting the histogram
plt.figure(figsize=(10, 6))              
df['Number'].plot(kind='hist', bins=5)    
plt.title('The title of the graph')       
plt.xlabel('The label of the x-axis')      
plt.ylabel('The label of the y-axis')    
plt.show()                                 
```

```python
# using Seaborn

import seaborn as sns
import pandas as pd

# Sample data
data = {
     'Number': [5,10,15,20,25,30,35,40,45,50]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))            
sns.histplot(df['Number'], bins=5)       
plt.title('The title of the graph')      
plt.xlabel('The label of the x-axis')     
plt.ylabel('The label of the y-axis')     
plt.show()                               
```
