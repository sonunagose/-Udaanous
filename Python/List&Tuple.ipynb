# Tuple Concatenation

```python
tuple_1 = (1,2,3,4,5)
tuple_2 = (2,3,4,5,6,)
result = tuple_1 + tuple_2
print(result)
```

# Nested Tuple Slicing

# Slicing A Nested Tuple

```python
nested_tuple = (("Chiku", "Mango"),(50,100,150),("A", "B", "C", "D"))
slicing_result = nested_tuple[1][0:2]
print(slicing_result)
```

```python
nested_tuple = (("Chiku", "Mango"),(50,100,150),("A", "B", "C", "D"))
slicing_result = nested_tuple[1][2]
print(slicing_result)
```

# List

```python
Number = [10,20,30,40,50,60,70]
print(Number)
```

# List Forward Slicing

```python
Fruits = ["Apple", "Banana", "Mango", "Chiku"]
forward_slicing = Fruits[1:3]
print(forward_slicing)
```

# List Backward Slicing

```python
Fruits = ["Apple", "Banana", "Mango", "Chiku"]
Backword_slicing = Fruits[-3:-1]
print(Backword_slicing)
```

# List Striding

```python
Number = [5,10,15,20,30,40,60,55]
Striding_range = Number[0:5:3]
print(Striding_range)
```

# List Item Add/Remove

### Adding Remove Value in List
