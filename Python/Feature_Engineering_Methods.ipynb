# 43.Creating Total Purchase Amount Variable

```python
# Mathematical Formula:
Total purchase amount = Purchase frequency * average purchase amount
```

```python
import pandas as pd

data['Total_Purchase_Amount'] = data['quantity'] * data['price']

print(data[['order_id', 'product_name', 'quantity', 'price', 'Total_Purchase_Amount']].head())
```

# Creating Customer Lifetime Value Variable

```python
# Mathematical Formula:
Customer lifetime value = Customer’s Value * Customer’s lifespan
```

```python
import pandas as pd
import numpy as np

np.random.seed(42) 
data['Customer_Lifespan_Months'] = np.random.randint(6, 36, size=len(data))

data['Total_Purchase_Amount'] = data['quantity'] * data['price']

data['CLV'] = data['Total_Purchase_Amount'] * data['Customer_Lifespan_Months']

print(data[['customer_id', 'Total_Purchase_Amount', 'Customer_Lifespan_Months', 'CLV']].head())
```

# 44.Date & Time

```python
data['order_date'] = pd.to_datetime(data['order_date'])

data['year'] = data['order_date'].dt.year
data['month'] = data['order_date'].dt.month
data['day'] = data['order_date'].dt.day

print(data[['order_date', 'year', 'month', 'day']].head())
```

# 45.Feature Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_columns = ['product_name', 'category', 'region']

for col in categorical_columns:
    data[col + '_encoded'] = le.fit_transform(data[col])

print(data[[*categorical_columns, *(col + '_encoded' for col in categorical_columns)]].head())

```

# 46.Feature Mapping

```python
region_mapping = {
    'North': 0,
    'South': 1,
    'East': 2,
    'West': 3
}

data['region_mapped'] = data['region'].map(region_mapping)

print(data[['region', 'region_mapped']].head())
```

# 46.Feature Bining

```python
import pandas as pd

data['Total_Purchase_Amount'] = data['quantity'] * data['price']

bins = [0, 100, 300, float('inf')]
labels = ['Low', 'Medium', 'High']

data['Purchase_Amount_Bin'] = pd.cut(data['Total_Purchase_Amount'],
                                     bins=bins,
                                     labels=labels,
                                     include_lowest=True)

print(data[['Total_Purchase_Amount', 'Purchase_Amount_Bin']].head())
```

# 47.Creating Dummies

```python
import pandas as pd

dummies = pd.get_dummies(data[['category', 'region']], prefix=['category', 'region'])

data = pd.concat([data, dummies], axis=1)

print(data.head())
```
