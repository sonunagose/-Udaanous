# 7.K-Means Elbow Methods

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

data = pd.read_excel('C:/Users/JOHN/Desktop/practice_data.xlsx')

label_enc = LabelEncoder()
data['gender'] = label_enc.fit_transform(data['gender'])
data['department'] = label_enc.fit_transform(data['department'])
data['Accuracy of Work'] = label_enc.fit_transform(data['Accuracy of Work'])  
data['Meeting Project Deadlines'] = label_enc.fit_transform(data['Meeting Project Deadlines']) 
data['churned'] = label_enc.fit_transform(data['churned'])

features = ['Days Present', 'Accuracy of Work', 'Meeting Project Deadlines',
            'Task Completion Rates', 'Interpersonal Skills Rating',
            'Decision-Making Skills Rating', 'Sales Revenue Generated']

X = data[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plotting the Elbow
plt.figure(figsize=(8, 6))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Method to Determine Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.grid(True)
plt.show()
```

# 8.Developing K-Means

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

data = pd.read_excel('C:/Users/JOHN/Desktop/practice_data.xlsx')

label_enc = LabelEncoder()
data['gender'] = label_enc.fit_transform(data['gender']) 
data['department'] = label_enc.fit_transform(data['department'])
data['Accuracy of Work'] = label_enc.fit_transform(data['Accuracy of Work']) 
data['Meeting Project Deadlines'] = label_enc.fit_transform(data['Meeting Project Deadlines'])
data['churned'] = label_enc.fit_transform(data['churned']) 

features = ['Days Present', 'Accuracy of Work', 'Meeting Project Deadlines',
            'Task Completion Rates', 'Interpersonal Skills Rating',
            'Decision-Making Skills Rating', 'Sales Revenue Generated']

X = data[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

selected_k_value = 3
kmeans = KMeans(n_clusters=selected_k_value, init='k-means++', random_state=42)
kmeans.fit(X_scaled)

#cluster labels
labels = kmeans.labels_


data['Cluster'] = labels

print(data[['EmployeeID', 'Cluster']])
```
