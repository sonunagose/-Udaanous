```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel("C:/Users/JOHN/Desktop/boxplot_data.xlsx")
```

```python
data
```

# Create a boxplot for the Task Completion Rates by churned

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='churned', y='Task Completion Rates', data=data)
plt.title('Task Completion Rates by Churned Status')
plt.xlabel('Churned')
plt.ylabel('Task Completion Rates')
plt.show()
```

# What is the median Task Completion Rates for the churned employees?

```python
median_churned = data[data['churned'] == 'Yes']['Task Completion Rates'].median()
print("Median Task Completion Rates for churned employees:", median_churned)
```
