# Step 1.Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Step 2.Load Data

```python
data = pd.read_excel('C:/Users/JOHN/Desktop/Customer Purchase Behavior Analysis Project.xlsx')
```

```python
data
```

# Step 3. Initial Data Exploration

```python
data.info()
```

```python
# View the first 5 rows
data.head()

# Checking for missing values
data.isnull().sum()

# Statistical summary
data.describe(include='all')
```

#  Step 4.Visualize Gender Distribution

```python
sns.countplot(data=data, x='Gender')
plt.title('Gender Distribution')
plt.show()
```

# Step 5: Analyze Spending Score by Gender

```python
# Using boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(data=data, x='Gender', y='SpendingScore')
plt.title('Spending Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Spending Score')
plt.grid(True)
plt.show()

# Average spending score by gender
avg_spending_by_gender = data.groupby('Gender')['SpendingScore'].mean()
print("Average Spending Score by Gender:\n", avg_spending_by_gender)
```

# Step 6.Explore Relationship Between Age and Spending Score

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='Age', y='SpendingScore', hue='Gender')
plt.title('Relationship Between Age and Spending Score')
plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.grid(True)
plt.show()
```

# Step 7.Category-Wise Spending Analysis

```python
category_avg_spending = data.groupby('PurchaseCategory')['SpendingScore'].mean().sort_values(ascending=False)

# Plot the average spending score by category
plt.figure(figsize=(8, 6))
category_avg_spending.plot(kind='bar', color='purple')
plt.title('Average Spending Score by Purchase Category')
plt.xlabel('Purchase Category')
plt.ylabel('Average Spending Score')
plt.show()

print("Average Spending Score by Category:\n", category_avg_spending)
```

# Step 8.More Analysis

```python
# Compare Income vs Spending

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='AnnualIncome', y='SpendingScore', hue='Gender')
plt.title('Annual Income vs Spending Score')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.grid(True)
plt.show()
```

```python
# Create Pie Chart for Category Distribution

category_distribution = data['PurchaseCategory'].value_counts()

plt.figure(figsize=(6, 6))
category_distribution.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Category Distribution')
plt.show()

```

```python
# Insights:-
- Spending score generally increases with income but may plateau at higher income levels.
- Electronics is the most popular spending category, followed by Groceries and Clothing.
- Marketing strategies could be tailored by income groups and preferred product categories.
```

# Step 9.Conclusion

```python
# Conclusion

* Who are the high spenders?**  
=> The high spenders tend to be individuals with higher annual incomes and are most often found in the older age groups. Specifically, customers aged 40 and above with annual incomes in the upper quartile stand out as the top spenders.

* Are there patterns by gender or age?**  
=> Yes, there are clear patterns. Age shows a strong correlation with spending, with middle-aged and older individuals typically spending more. Gender may also play a role, depending on the category of products, though differences are less pronounced than those seen by age or income.

* What could a marketing team do based on this?**  
=> The marketing team could tailor campaigns to target high-income, older customers with premium offerings. Additionally, segmenting marketing strategies by age groups—offering loyalty programs or targeted promotions—could improve engagement. Gender-based product recommendations or ads could also be optimized based on spending tendencies.

```
