# Expersion & Variable

```python
A person buys eggs over three days. On Day 1, they buy 12 eggs; on Day 2, they buy 10 eggs; and on Day 3, they buy 8 eggs.

### Over the three days, they use the following number of eggs:
1. On Day 1, they use 5 eggs.
2. On Day 2, they use 6 eggs.
3. On Day 3, they use 4 eggs.
```

# The total number of eggs bought over the three days

```python
eggs_day1 = 10
eggs_day2 = 12
eggs_day3 = 5

total_eggs_bought = eggs_day1 + eggs_day2 + eggs_day3

print("Total number of eggs bought:", total_eggs_bought)
```

# The total number of eggs used over the three days

```python
used_day1 = 5
used_day2 = 6
used_day3 = 4

total_eggs_used = used_day1 + used_day2 + used_day3

print("Total number of eggs used over the three days :", total_eggs_used)
```

# The number of remaining eggs after using them

```python
remaining_eggs = total_eggs_bought - total_eggs_used

print("Number of Remaining of eggs brought:", remaining_eggs)
```
