# Imputing Missing Value

```python
import pandas as pd
from sklearn.impute import SimpleImputer

# Sample data with a missing value
data = pd.DataFrame({
    'ID': [10, 15, None, 20, 25]
})

# Create the imputer with mean strategy
imputer = SimpleImputer(strategy='mean')

# Fit the imputer and transform the data
imputer.fit(data[['ID']])
data[['ID']] = imputer.transform(data[['ID']])

# Show the result
print(data)
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer

# Sample data with a missing value
data = pd.DataFrame({
    'Age': [35, 45, 25,None, 20, 25]
})

# Create the imputer with mean strategy
imputer = SimpleImputer(strategy='mean')

# Fit the imputer and transform the data
imputer.fit(data[['Age']])
data[['Age']] = imputer.transform(data[['Age']])

# Show the result
print(data)
```

# 13.Combining & Conditions

```python
import pandas as pd

data = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 22, 28],
    'salary' :[10000,20000,30000,40000]
})
## OR Condition
filtered_data = data[(data['age'] > 25) | (data['salary'] < 50000)]
filtered_data
```

```python
## AND Condition
filtered_data = data[(data['age'] > 25) & (data['salary'] < 50000)]
filtered_data
```

```python

# Filter: name starts with 'A' AND age is less than 30
filtered_data = data[(data['name'].str.startswith('A')) & (data['age'] < 30)]
print(filtered_data)
```

# 14.Joining Data

```python
import pandas as pd

# First DataFrame
data1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

# Second DataFrame
data2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'Score': [85, 92, 78, 88]
})
inner_join = pd.merge(data1, data2, on='ID', how='inner')
print(inner_join)
```

# 15.Concatenating Data

```python
import pandas as pd

# First DataFrame
data1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Alice', 'Bob']
})

# Second DataFrame (same columns as data1)
data2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Charlie', 'David']
})

# Concatenate vertically
concatenated_data = pd.concat([data1, data2])

# Display result
print(concatenated_data.head())
```

# 18.Group By

```python
import pandas as pd

# Dataframe
data = pd.DataFrame({
    'ID' : [1,2,3,4,5],
    'Name': ['Alice', 'Bob', 'Charlie', 'David','Eve']
})
# Group by
Grouped = data.groupby('Name')['ID'].mean()

print(Grouped)
    
```

# 19.Pivot Table

```python
import pandas as pd

# Sample data
data = pd.DataFrame({
    'ID' : [1,2,3,4,5],
    'Age' : [20,25,30,35,40],
    'Name': ['Alice', 'Bob', 'Charlie', 'David','Eve']
})

# Pivot table
pivot_table = data.pivot_table(
    index='Name',
    values=['ID', 'Age'],
    aggfunc={'ID': 'mean', 'Age': 'mean'}
)

print(pivot_table)
```

# 20.Cross Tab

```python
import pandas as pd

# Sample data
data = pd.DataFrame({
    'Number_1': ['1', '1', '2', '2', '3', '1', '3'],
    'Alpha': ['X', 'Y', 'X', 'Y', 'Y', 'X', 'X']
})

# Crosstab with totals and normalized row-wise
cross_tab = pd.crosstab(
    data['Number_1'],
    data['Alpha'],
    margins=True,             
    normalize='index'          
)

print(cross_tab)
```
