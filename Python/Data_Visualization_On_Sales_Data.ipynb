# Line Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

data = {
    'order_id': [1001, 1002, 1003, 1004, 1005, 1196, 1197, 1198, 1199, 1200],
    'customer_id': [521, 509, 503, 511, 507, 520, 501, 504, 521, 529],
    'product_id': ['A101', 'A102', 'A102', 'D401', 'C302', 'B201', 'B202', 'A102', 'C302', 'A102'],
    'product_name': ['Wireless Mouse', 'Keyboard', 'Keyboard', 'Water Bottle', 'Pen Set', 
                     'Desk Lamp', 'Bookshelf', 'Keyboard', 'Pen Set', 'Keyboard'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Stationery',
                 'Furniture', 'Furniture', 'Electronics', 'Stationery', 'Electronics'],
    'quantity': [4, 3, 5, 2, 3, 3, 4, 5, 3, 3],
    'price': [25.0, 45.0, 45.0, 10.0, 5.0, 30.0, 120.0, 45.0, 5.0, 45.0],
    'order_date': ['2024-10-15', '2024-12-11', '2024-10-21', '2024-12-22', '2024-12-14',
                   '2024-11-24', '2024-10-16', '2024-10-30', '2024-10-17', '2024-11-27'],
    'region': ['North', 'West', 'North', 'South', 'West', 'West', 'South', 'West', 'North', 'South']
}
df = pd.DataFrame(data)

# Convert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'])

df['total_sales'] = df['quantity'] * df['price']

# Extract month 
df['Month'] = df['order_date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['total_sales'].sum().reset_index()
monthly_sales['Month'] = monthly_sales['Month'].astype(str)

plt.figure(figsize=(10, 6))  
plt.plot(monthly_sales['Month'], monthly_sales['total_sales'], marker='o') 
plt.title('Monthly Total Sales')
plt.xlabel('Month')  
plt.ylabel('Total Sales (USD)')  
plt.grid(True)
plt.show() 
```

#  Clustered (Grouped) Bar Chart: Sales by Region and Category

```python
import pandas as pd
import matplotlib.pyplot as plt

# [Assuming df is already created and cleaned as shown before]
df['order_date'] = pd.to_datetime(df['order_date'])
df['total_sales'] = df['quantity'] * df['price']

# Group by region and category
grouped = df.groupby(['region', 'category'])['total_sales'].sum().unstack()

# Clustered Bar Chart
plt.figure(figsize=(10, 6))
grouped.plot(kind='bar')  # clustered by default
plt.title('Total Sales by Region and Category')
plt.xlabel('Region')
plt.ylabel('Total Sales (USD)')
plt.xticks(rotation=0)
plt.legend(title='Category')
plt.tight_layout()
plt.show()
```

# Bar Chart

```python
import pandas as pd
import matplotlib.pyplot as plt

# columns like: 'category', 'quantity', 'price

# Calculate total sales
df['total_sales'] = df['quantity'] * df['price']

# Group by category
category_sales = df.groupby('category')['total_sales'].sum().sort_values(ascending=False)

# Plot bar chart
plt.figure(figsize=(10, 6))  
category_sales.plot(kind='bar', color='green') 

# Add
plt.title('Total Sales by Category')  
plt.xlabel('Product Category')        
plt.ylabel('Total Sales (USD)')      
plt.xticks(rotation=45)              
plt.show()                      
```

# Pie Chart

```python
import pandas as pd
import matplotlib.pyplot as plt

df['total_sales'] = df['quantity'] * df['price']

# Group by total sales by category
category_sales = df.groupby('category')['total_sales'].sum()

# pie chart
plt.figure(figsize=(10, 6))  
category_sales.plot(kind='pie', autopct='%1.1f%%', startangle=140)

plt.title('Sales Distribution by Product Category')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

# Histogram Chart

```python
# using Matplot

import matplotlib.pyplot as plt

# total_sales is calculated
df['total_sales'] = df['quantity'] * df['price']

# Plot histogram
plt.figure(figsize=(10, 6)) 
df['total_sales'].plot(kind='hist', bins=5, color='Orange', edgecolor='black')
plt.title('Distribution of Total Sales') 
plt.xlabel('Total Sales (USD)')         
plt.ylabel('Frequency')                  
plt.show()
```

```python
# Using Seaborn
import seaborn as sns

sns.histplot(df['total_sales'], bins=5, kde=True, color='orange')
plt.title('Distribution of Total Sales')
plt.xlabel('Total Sales (USD)')
plt.ylabel('Count')
plt.show()
```

# 29.Heatmap

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.DataFrame([
    [1001, 521, 'A101', 'Wireless Mouse', 'Electronics', 4, 25.0, '2024-10-15', 'North'],
    [1002, 509, 'A102', 'Keyboard', 'Electronics', 3, 45.0, '2024-12-11', 'West'],
    [1003, 503, 'A102', 'Keyboard', 'Electronics', 5, 45.0, '2024-10-21', 'North'],
    [1004, 511, 'D401', 'Water Bottle', 'Accessories', 2, 10.0, '2024-12-22', 'South'],
    [1005, 507, 'C302', 'Pen Set', 'Stationery', 3, 5.0, '2024-12-14', 'West'],
    [1197, 501, 'B202', 'Bookshelf', 'Furniture', 4, 120.0, '2024-10-16', 'South'],
    [1198, 504, 'A102', 'Keyboard', 'Electronics', 5, 45.0, '2024-10-30', 'West'],
    [1199, 521, 'C302', 'Pen Set', 'Stationery', 3, 5.0, '2024-10-17', 'North'],
    [1200, 529, 'A102', 'Keyboard', 'Electronics', 3, 45.0, '2024-11-27', 'South']
], columns=[
    'order_id', 'customer_id', 'product_id', 'product_name', 'category',
    'quantity', 'price', 'order_date', 'region'
])

data['order_date'] = pd.to_datetime(data['order_date'])

numeric_data = data.select_dtypes(include=['float64', 'int64'])

correlation_matrix = numeric_data.corr()

value_width = 8
value_height = 6

# Plot heatmap
plt.figure(figsize=(value_width, value_height))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Sales Data')
plt.show()
```
